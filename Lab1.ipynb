```python
# write a program to create a numpy array using a python list.
# display the array and print the shape and datatype of array.

import numpy as np

arr1 = np.array([1, 2, 3, 4, 5])
print(arr1)
print("shape: ", arr1.shape)
print("datatype: ", arr1.dtype)
```

```python
# write a program to create a 3x3 matrix of 0s and create a 2x5 matrix of 1s.
# print both the matrices and their shape.

matrix0 = np.zeros((3,3))
matrix1 = np.ones((2,5))

print("matrix 1 shape: ", matrix0.shape)
print(matrix0)

print("\nmatrix 2 shape: ", matrix1.shape)
print(matrix1)
```

```python
# write a program to create a numpy array of numbers from 1 to 12.
# reshape the array into a 3x4 matrix. display the original and reshaped
# array along with their shapes.

og_arr = np.arange(1, 13)
print("original array shape: ", og_arr.shape)
print(og_arr)

reshaped_arr = og_arr.reshape(3,4)
print("\nreshaped array shape: ", reshaped_arr.shape)
print(reshaped_arr)
```

```python
# write a program to create a numpy array of 10 random integers.
# find the minimum, maximum, mean and sum of elements of the array.

arr4 = np.random.randint(1, 100, 10)
print(arr4)
print("minimum: ", arr4.min())
print("maximum: ", arr4.max())
print("mean: ", arr4.mean())
print("sum: ", arr4.sum())
```

```python
# write a program to create a dataframe using a dictionary containing student name and marks.
# display the data frame and print the datatypes of each column.

import pandas as pd
data = {
    "name": ["Gayathri", "Ansh", "Vedica", "Ishika"],
    "marks": [100, 100, 99, 69]
}

df = pd.DataFrame(data)

print(df)
print("\ndatatypes: ")
print(df.dtypes)
```
